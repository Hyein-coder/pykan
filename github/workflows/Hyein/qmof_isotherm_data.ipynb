# IsothermNet Dataset Inspection

CO2 uptake and heat of adsorption in 19 distinct pressures points

```python
data_pressure = [1, 5, 10, 50, 100, 200, 300, 400, 500, 700, 10000, 1500, 2000, 2500, 3000, 3500, 4000, 4500, 5000] # kPa
```

```python
import pandas as pd
data_txt = pd.read_excel(r'../../../.ignore/isothermnet_full_dataset/texturalProperties_vol.xlsx')
data_txt.head()
```

```python
import torch

# Load the data into a variable
data_x = torch.load('../../../.ignore/isothermnet_full_dataset/X_dataset_electro_xyz_bond_struc.pth')
data_y = torch.load('../../../.ignore/isothermnet_full_dataset/y_dataset19.pth')
data_H = torch.load('../../../.ignore/isothermnet_full_dataset/H_dataset.pth')

dataset = (data_x, data_y, data_H)
dataset_name = ['X', 'Y', 'H']
for d, name in zip(dataset, dataset_name):
    print(f"================ {name} ================")
    for k in d.keys():
        print(f"{k} Size: {d[k].shape}")
```

```python
import os

os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import numpy as np

output_name = 'heat'
x_vals = data_H['H'][:,:].numpy()  # (5394, 19)
n_cat = x_vals.shape[1]

txt_vals = data_txt.iloc[:,:].values

fig, axes = plt.subplots(2, 4, figsize=(12,6))
axes = axes.flatten()

cmap = plt.get_cmap('viridis', n_cat)

for idx in range(7):
    ax = axes[idx]
    feature = txt_vals[:, idx+1]

    x_all = np.tile(feature, n_cat)
    y_all = x_vals.flatten('F')
    c_all = np.repeat(np.arange(n_cat), len(feature))

    ax.scatter(x_all, y_all, c=c_all, cmap='viridis', vmin=0, vmax=n_cat-1,
               s=1, alpha=0.3, linewidths=0, rasterized=True)
    ax.set_xlabel(data_txt.columns[idx+1])
    ax.set_ylabel("CO₂ Uptake")

# Use the last (8th) subplot for the legend
legend_ax = axes[-1]
legend_ax.set_visible(True)
legend_ax.axis('off')

legend_handles = [
    mpatches.Patch(color=cmap(i / (n_cat - 1)), label=f'{data_pressure[i]} kPa')
    for i in range(n_cat)
]

legend_ax.legend(
    handles=legend_handles,
    title="Pressure",
    loc='center',
    fontsize=8,
    title_fontsize=9,
    ncol=2 if n_cat > 10 else 1,
    frameon=False
)

plt.tight_layout()
plt.savefig(f"./data/isotherm/{output_name}_scatter.png", dpi=120)
print("Done!")
```

```python
import os

os.environ["KMP_DUPLICATE_LIB_OK"] = "TRUE"

import matplotlib.pyplot as plt
import numpy as np

p_idx = 17
p_val = data_pressure[p_idx]

# output_name = 'heat'
# output_label = 'H_ads [kJ/mol]'
# y_atP = data_H['H'][:,p_idx:p_idx+1].numpy()  # (5394, 19)

output_name = 'isotherm'
output_label = 'CO₂ Uptake [g/g]'
y_atP = data_y['isotherm'][:,p_idx:p_idx+1].numpy()  # (5394, 19)

feat_vals = data_txt.iloc[:,1:-1].values
feat_names = data_txt.columns.tolist()[1:-1]

fig, axes = plt.subplots(3, 2, figsize=(8, 12))
axes = axes.flatten()

for idx in range(feat_vals.shape[1]):
    ax = axes[idx]
    feat_i = feat_vals[:,idx]

    ax.scatter(feat_i, y_atP,
               s=1, alpha=0.3, linewidths=0, rasterized=True)
    ax.set_xlabel(feat_names[idx])
    ax.set_ylabel(output_label)

# axes[-1].set_visible(False)
plt.tight_layout()
plt.savefig(f"./data/isotherm/{output_name}_scatter_at_{p_val}.png", dpi=120)
print(f"Done!: p = {p_val}")
```

```python
data_to_save = np.concatenate([feat_vals, y_atP], axis=1)
names_to_save = feat_names+[output_label]

df_to_save = pd.DataFrame(data_to_save, columns=names_to_save)
print(df_to_save.head())
if output_name == 'heat':
    file_tag = 'CO2AH'
else:
    file_tag = 'CO2AR'

df_to_save.to_csv(f"./data/{file_tag}{p_val}.csv", index=False)
print(f'Saved as "./data/{file_tag}{p_val}.csv"')
```

## Combined dataset — all pressures, both outputs

One big table over all 19 pressure points. Each row pairs the 6 textural features with the **Pressure** value (as a feature) and both targets at that pressure: **H_ads** and **CO₂ Uptake**. Total rows = 5394 MOFs × 19 pressures = 102,486.

```python
# ---- Build one big dataset: all 19 pressures, both outputs (H_ads & CO2 Uptake) ----
# Each row = [6 textural features, Pressure, H_ads, CO2 Uptake]
# Pressure is included as an explicit feature column.
# Rows ordered pressure-major: all MOFs at pressure[0], then all MOFs at pressure[1], ...

feat_vals = data_txt.iloc[:, 1:-1].values        # (5394, 6) -- drops MOF id and No. Channels
feat_names = data_txt.columns.tolist()[1:-1]

H_all = data_H['H'].numpy()                      # (5394, 19) -- heat of adsorption
uptake_all = data_y['isotherm'].numpy()          # (5394, 19) -- CO2 uptake
n_mof, n_p = H_all.shape                          # (5394, 19)

# Tile features once per pressure (pressure-major block stacking)
feat_tiled = np.tile(feat_vals, (n_p, 1))                                   # (5394*19, 6)
pressure_col = np.repeat(np.asarray(data_pressure, dtype=float), n_mof)[:, None]  # (5394*19, 1)
H_col = H_all.flatten('F')[:, None]              # column-major -> pressure-major
uptake_col = uptake_all.flatten('F')[:, None]    # column-major -> pressure-major

# ---- Randomly subsample rows to reduce dataset size ----
# Same row indices are used for both outputs so AIH and AIR stay aligned.
n_sample = 5000                                  # target number of rows (out of n_mof * n_p)
rng = np.random.default_rng(seed=42)             # fixed seed for reproducibility
n_total = n_mof * n_p
sample_idx = rng.choice(n_total, size=n_sample, replace=False)
sample_idx.sort()                                # keep pressure-major ordering

config_datasets = {
    'AIH': ['H_ads [kJ/mol]', H_col],
    'AIR': ['CO2_uptake [g/g]', uptake_col]
}

for name, col_info in config_datasets.items():
    col_name, col_data = col_info
    big_data = np.concatenate([feat_tiled, pressure_col, col_data], axis=1)
    big_data = big_data[sample_idx]              # apply random subsample
    big_names = feat_names + ['Pressure [kPa]', col_name]
    df_big = pd.DataFrame(big_data, columns=big_names)
    print(f"Combined dataset shape: {df_big.shape}  (subsampled from {n_total} = {n_mof} MOFs x {n_p} pressures)")
    print(df_big.head())

    out_path = f"./data/{name}.csv"
    df_big.to_csv(out_path, index=False)
    print(f'Saved as "{out_path}"')
```
